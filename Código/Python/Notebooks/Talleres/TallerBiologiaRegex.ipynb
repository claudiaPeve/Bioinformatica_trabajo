#### 5.- De la lista de accessions que se proporciona obtén lo siguiente:
1. Los que tengan el número 5
2. Los que tengan la letra 'd' o la 'e'
3. Los que tengan las letras 'd' y 'e' en ese orden
4. Los que tengan las letras 'd' y 'e' en ese orden con una letra cualquiera en medio
5. Los que tengan ambas letras 'd' y 'e' en cualquier orden
6. Los que empiezan por 'x' o por 'y'
7. Los que empiezan por 'x' o 'y' y terminan por 'e'
8. Los que tienen tres o más dígitos seguidos
9. Los que terminan con 'd' seguidos de 'a' o 'r' o 'p'

```python
import re
```

```python
accs = ['xkn59438', 'yhdck2', 'eihd39d9', 'chdsye847', 'hedle3455', 'xjhd53e', '45da', 'de37dp']
```

```python
for elemento in accs:
    if re.search(r'5',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'd',elemento) or re.search(r'e',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'd|e',elemento):
        print(elemento)
```

```python
#si queremos que no se raye con mayusculas y minusculas podemos convertir con .lower o .upper antes de leer, si no hay que incluirlo en el argumento
for elemento in accs:
    if re.search(r'[de]',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'de',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'd.*e',elemento):
        print(elemento)
```

```python
#así me obligo a que haya algo en medio
for elemento in accs:
    if re.search(r'd.+e',elemento):
        print(elemento)
```

```python
#así una d o e, algo o no y otra d o e
for elemento in accs:
    if re.search(r'[de].*[de]',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'd',elemento) and re.search(r'd',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'^x',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'^x|^y',elemento):
        print(elemento)
```

```python
#si el sombrerito esta fuera de los corchetes significa inicio de línea o elemento, si va dentro es negación
for elemento in accs:
    if re.search(r'^[xy]',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'^[xy].*e$',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'\d{3}',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'\d{3,}',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'[0-9]{3}',elemento):
        print(elemento)
```

```python
for elemento in accs:
    if re.search(r'd[arp]$',elemento):
        print(elemento)
```

```python
#forma de hacerlo condensado pero a veces el segundo no se reconoce con símbolos especiales
for elemento in accs:
    if re.search(r'd' or 'e',elemento):
        print(elemento)
```
