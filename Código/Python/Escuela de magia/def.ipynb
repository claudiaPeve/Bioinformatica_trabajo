```python
def escuela_de_magia():
    print ("Bienvenid@ pequeño ser de luz✨, ha llegado la convocatoria oficial de los Exámenes Mágicos Supremos📝, ¿deseas presentarte?")
    while True:
        respuesta = input(" Escribe 1 para continuar✅, escribe 2 para salir del juego❌:")
        #respuesta = int(respuesta)
        if respuesta == "1" or respuesta == "2":
            if respuesta == "1":
                print("\nSabia decisión, el rector👴🏿 te leerá las reglas de los exámenes a continuación:\n")
                print("Eres un pequeño mago/a del reino de Alfea🪷, acabas de cumplir 18 años y te has matriculado en la Escuela Superior de Magia.\nDurante 3 años te has estado formando para convertirte en un/a gran mago/a de tu reino.\nEstás en tus exámenes finales, dependiendo de la nota que saques al final te graduarás como: Brujo/a, Hadito/a o Cambiaformas")
                print("\n\n\033[1m-----NIVEL 1: EXAMEN DE POCIONES-----\033[0m. \n\nEl primer examen es de pociones. Lo realizarás en el invernadero de la Bruja Suprema Asterin Blackbeak🧙🏼‍♀️. \nTendrás que elegir los ingredientes correctos para elaborar la poción para borrar recuerdos🔮.\n\n")
                print("Aquí tienes la receta incompleta, faltan 2 ingredientes: \n\n- Agua de luna🌙\n________________\n- Lágrimas de Fénix🐦‍🔥\n__________________\n- Escamas de serpiente dorada🐍\n\n")
                print("Estas son las opciones que tienes para completarla:\n1) Raíz de la amnesia🪾\n2) Polvo de Cronos🕐\n3) Plumas de grifo🪶\n4) Seda del Reino de las Arañas🕷️\n\nIndica el número del ingrediente correcto sin repetir.\n")
                while True:
                    lista_intentos_escobas = []
                    puntuacion_e=0
                    import re
                    intento_escoba1 = input("Indica el número correspondiente al primer ingrediente1️⃣:")
                    if intento_escoba1 in "1,2,3,4":
    
                        while True:
                            intento_escoba2 = input("Indica el número correspondiente al segundo ingrediente2️⃣:")
                            if intento_escoba2 in "1,2,3,4":
                                lista_intentos_escobas += [intento_escoba1,intento_escoba2]
                                
                                if re.search(r"1",lista_intentos_escobas[0]) and re.search(r"2",lista_intentos_escobas[1]):
                                    puntuacion_e=puntuacion_e+3
                                    print("\n\033[1mBIEN, tienes 3/3 escobas🧹🧹🧹\033[0m\n")
                                    break
                                elif re.search(r"2",lista_intentos_escobas[0]) and  re.search(r"1",lista_intentos_escobas[1]):
                                    puntuacion_e=puntuacion_e+3
                                    print("\n\033[1mBIEN, tienes 3/3 escobas🧹🧹\033[0m\n")
                                    break
                                elif re.search(r"1|2",lista_intentos_escobas[0]) and  re.search(r"[^12]",lista_intentos_escobas[1]):
                                    puntuacion_e=puntuacion_e+2
                                    print("\n\033[1mCASI, tienes 2/3 escobas🧹🧹\033[0m\n")
                                    break
                                elif re.search(r"1|2",lista_intentos_escobas[1]) and  re.search(r"[^12]",lista_intentos_escobas[0]):
                                    puntuacion_e=puntuacion_e+2
                                    print("\n\033[1mCASI, tienes 2/3 escobas🧹🧹\033[0m\n")
                                    break
                                elif re.search(r"1",lista_intentos_escobas[1]) and  re.search(r"1",lista_intentos_escobas[0]):
                                    puntuacion_e=puntuacion_e+2
                                    print("\n\033[1mCASI, tienes 2/3 escobas🧹🧹\033[0m\n")
                                    break
                                elif re.search(r"2",lista_intentos_escobas[1]) and  re.search(r"2",lista_intentos_escobas[0]):
                                    puntuacion_e=puntuacion_e+2
                                    print("\n\033[1mCASI, tienes 2/3 escobas🧹🧹\033[0m\n")
                                    break
                                elif re.search(r"3|4",lista_intentos_escobas[1]) and  re.search(r"3|4",lista_intentos_escobas[0]):
                                    puntuacion_e=puntuacion_e+1
                                    print("\n\033[1mOh vaya, tienes solo 1/3 escobas🧹\033[0m\n")
                                    break
                                #else:
                                    #print("\nNo te entiendo pequeño ser de luz.🤨\n")
                                    #break
                            else:
                                print("\nNo te entiendo pequeño ser de luz.🤨\n")
                        break
                    else: 
                        print("\nNo te entiendo pequeño ser de luz.🤨\n")
                    
                print("\n\n\033[1m-----NIVEL 2: EXAMEN DE HECHIZOS-----\033[0m \n\nEl segundo examen es sobre hechizos, con tu profesor Paladium Everwood, el duende🧝🏼‍♂️ que imparte sus clases en el Jardín de los Secretos.\nTendrás que conjurar un hechizo 🪄para teletransportarte 🌀.\nTienes que rellenar las 3 palabras que faltan:")
                print("\n“Donde —----- No Quiero —---, Donde —---- No Puedo Llegar. Transitus Liminis”\n\n")
                print("Estas son las opciones que tienes para completarla:\n1) Estar \n2) Quiero \n3) Estoy\nIndica el número de la palabra correcta sin repetir.\n")
                while True:
                    lista_intentos_hechizo = []
                    puntuacion_v=0
                    intento_hechizo1 = input("Indica el número correspondiente a la primera palabra1️⃣:")
                    if intento_hechizo1 in "1,2,3":

                        while True:
                            intento_hechizo2 = input("Indica el número correspondiente a la segunda palabra2️⃣:")
                            if intento_hechizo2 in "1,2,3":
                                while True: 
                                    intento_hechizo3 = input("Indica el número correspondiente a la tercera palabra3️⃣:")
                                    if intento_hechizo3 in "1,2,3":
                                        lista_intentos_hechizo += [intento_hechizo1,intento_hechizo2,intento_hechizo3]
                                        
                                        if re.search(r"3",lista_intentos_hechizo[0]) and re.search(r"1",lista_intentos_hechizo[1]) and re.search(r"2",lista_intentos_hechizo[2]):
                                            puntuacion_v=puntuacion_v+3
                                            print("\n\033[1mBIEN, tienes 3/3 varitas🪄🪄🪄\033[0m")
                                            break
                                        elif re.search(r"3",lista_intentos_hechizo[0]) and  re.search(r"1",lista_intentos_hechizo[1]) and re.search(r"[^2]",lista_intentos_hechizo[2]):
                                            puntuacion_v=puntuacion_v+2
                                            print("\n\033[1mCASI, tienes 2/3 varitas🪄🪄\033[0m")
                                            break
                                        elif re.search(r"3",lista_intentos_hechizo[0]) and  re.search(r"[^1]",lista_intentos_hechizo[1]) and re.search(r"2",lista_intentos_hechizo[2]):
                                            puntuacion_v=puntuacion_v+2
                                            print("\n\033[1mCASI, tienes 2/3 varitas🪄🪄\033[0m")
                                            break
                                        elif re.search(r"[^3]",lista_intentos_hechizo[0]) and  re.search(r"1",lista_intentos_hechizo[1]) and re.search(r"2",lista_intentos_hechizo[2]):
                                            puntuacion_v=puntuacion_v+2
                                            print("\n\033[1mCASI, tienes 2/3 varitas🪄🪄\033[0m")
                                            break
                                        #elif re.search(r"[^123]",lista_intentos_hechizo[0]) and  re.search(r"[^123]",lista_intentos_hechizo[1]) and re.search(r"[^123]",lista_intentos_hechizo[2]):
                                            #print("\n\033[1mNo entiendo.\033[0m")
                                            #8break
                                        else: 
                                            puntuacion_v=puntuacion_v+1
                                            print("\n\033[1mOh vaya, tienes solo 1/3 varitas🪄\033[0m")
                                            break   
                                    else:
                                        print("\nNo te entiendo pequeño ser de luz🤨.\n")
                                break
                            else:
                                print("\nNo te entiendo pequeño ser de luz🤨.\n")
                        break
                    else:
                        print("\nNo te entiendo pequeño ser de luz🤨.\n")
                                #print(puntuacion_v)
                    
                print("\n\n\033[1m-----NIVEL 3: EXAMEN DE TRANSFORMACIONES-----\033[0m. \n\nEn último examen tienes que decidir en qué te transformarás para llevar a cabo distintas misiones. \nEste examen lo harás con tu profesora cambiaformas Violet Ennar🐆.")
                print("\nTendrás que transformarte en la criatura más adecuada para llevar a cabo cada misión. \nLas misiones son las siguientes:\n A) Llevar un mensaje cifrado al reino vecino📜. \n B) Encontrar las perlas escondidas en el fondo del Lago de las Sirenas🫧. \n C) Espiar al Capitán de la Guardia real💂🏻‍♀️.")
                print("\nA continuación aparecen tus opciones:\n 1) Cuervo🦅\n 2) Ratón🐀\n 3) Dragón marino🐲\nIndica el número de la criatura correcta sin repetir.\n")
                while True:
                    lista_intentos_camaleon = []
                    puntuacion_c=0
                    intento_camaleon1 = input("Indica el número correspondiente a la criatura para la misión A:")
                    if intento_camaleon1 in "1,2,3":
                        
                        while True:
                            intento_camaleon2 = input("Indica el número correspondiente a la criatura para la misión B:")
                            if intento_camaleon2 in "1,2,3":
                                while True:
                                    intento_camaleon3 = input("Indica el número correspondiente a la criatura para la misión C:")
                                    if intento_camaleon3 in "1,2,3":
                                        lista_intentos_camaleon += [intento_camaleon1,intento_camaleon2,intento_camaleon3]
                                        if re.search(r"1",lista_intentos_camaleon[0]) and re.search(r"3",lista_intentos_camaleon[1]) and re.search(r"2",lista_intentos_camaleon[2]):
                                            puntuacion_c=puntuacion_c+3
                                            print("\n\033[1mBIEN, tienes 3/3 camaleones🦎🦎🦎\033[0m")
                                            break
                                        elif re.search(r"1",lista_intentos_camaleon[0]) and  re.search(r"3",lista_intentos_camaleon[1]) and re.search(r"[^2]",lista_intentos_camaleon[2]):
                                            puntuacion_c=puntuacion_c+2
                                            print("\n\033[1mCASI, tienes 2/3 camaleones🦎🦎\033[0m")
                                            break
                                        elif re.search(r"1",lista_intentos_camaleon[0]) and  re.search(r"[^3]",lista_intentos_camaleon[1]) and re.search(r"2",lista_intentos_camaleon[2]):
                                            puntuacion_c=puntuacion_c+2
                                            print("\n\033[1mCASI, tienes 2/3 camaleones🦎🦎\033[0m")
                                            break
                                        elif re.search(r"[^1]",lista_intentos_camaleon[0]) and  re.search(r"3",lista_intentos_camaleon[1]) and re.search(r"2",lista_intentos_camaleon[2]):
                                            puntuacion_c=puntuacion_c+2
                                            print("\n\033[1mCASI, tienes 2/3 camaleones🦎🦎\033[0m")
                                            break
                                        #elif re.search(r"[^123]",lista_intentos_hechizo[0]) and  re.search(r"[^123]",lista_intentos_hechizo[1]) and re.search(r"[^123]",lista_intentos_hechizo[2]):
                                            #print("\nNo entiendo.🤨")
                                            #break
                                        else: 
                                            puntuacion_c=puntuacion_c+1
                                            print("\n\033[1mOh vaya, tienes solo 1/3 camaleones🦎")
                                            break
                                    #print(puntuacion_c)  
                                    else:
                                        print("\nNo te entiendo pequeño ser de luz.🤨\n")
                                break
                            else:
                                print("\nNo te entiendo pequeño ser de luz.🤨\n")
                        break
                    else:
                        print("\nNo te entiendo pequeño ser de luz.🤨\n")

                print("\n\n\033[1m-----FIN DE CURSO-----\033[1m\n\nFelicidades, has terminado tus exámenes🎉🎊. Los maestros se reunirán y decidirán tu destino.\nSi has obtenido más escobas🧹 te graduarás como bruj@🧙🏻‍♀️.\nSi has obtenido más varitas🪄 te graduarás como hadit@🧚🏼‍♀️.\nSi has obtenido más camaleones🦎te graduarás como cambiaformas🪅.\nSi no alcanzas una nota igual o superior a 5 entre todos, debes repetir el curso🔙.\n")
                from IPython.display import Image, display
                while True:
                    nota = puntuacion_e + puntuacion_v + puntuacion_c
                    if nota >= 5:
                        if puntuacion_e > puntuacion_v and puntuacion_e > puntuacion_c:
                            print("\n\033[1m¡ENHORABUENA! Te has convertido en bruj@\033[0m🧙🏻‍♀️.")
                            display (Image("HOCUS-POCUS.webp", width=250))
                            break
                        elif puntuacion_v > puntuacion_e and puntuacion_v > puntuacion_c:
                            print("\n\033[1m¡ENHORABUENA! Te has convertido en hadit@\033[0m🧚🏼‍♀️.")
                            display (Image("Clarion.jpg", width=125))
                            break   
                        elif puntuacion_c > puntuacion_e and puntuacion_c > puntuacion_v:
                            print("\n\033[1m¡ENHORABUENA! Te has convertido en cambiaformas\033[0m🪅.")
                            display (Image("cambiaformas2.jpg", width=125))
                            break
                        elif puntuacion_v == puntuacion_e or puntuacion_v == puntuacion_c or puntuacion_e == puntuacion_c:
                            print("\nVaya, parece que tienes dotes magníficas de varias disciplinas🤭 y los maestros no lo tienen claro.\nPara salir de dudas, elige un familiar🫶🏻 que te acompañará el resto de tu vida profesional mágica.\nTus poderes se mezclarán con los de tu familiar ¡Elige con cabeza! :\n\n1) Gato negro🐈‍⬛\n2) Búho🦉\n3) Serpiente alada🐍.\n")
                            while True:
                                familiar = input("Indica el número del familiar elegido: ")
                                if familiar == "1":
                                    print("\n\033[1m¡ENHORABUENA! Te has convertido en bruj@🧙🏻‍♀️\033[0m.")
                                    display (Image("HOCUS-POCUS.webp", width=250))
                                    break
                                elif familiar == "2":
                                    print("\n\033[1m¡ENHORABUENA! Te has convertido en hadit@🧚🏼‍♀️\033[0m.")
                                    display (Image("Clarion.jpg", width=125))
                                    break
                                elif familiar == "3":
                                    print("\n\033[1m¡ENHORABUENA! Te has convertido en cambiaformas🪅\033[0m.")
                                    display (Image("cambiaformas2.jpg", width=125))
                                    break
                                else:
                                    print("\nNo te entiendo🤨")
                    else:
                        print("Vaya, parece que aun no estás preparado para graduarte. Vuelve a intentarlo el curso que viene🧐.")
                        break
                    break
            else:
                print ("Vaya!, parece que aun no estás preparad@, nos veremos en la próxima convocatoria➰")
                break
            break
    else:
        print("No entiendo tu respuesta, vuelve a intentarlo🤨")               
```
