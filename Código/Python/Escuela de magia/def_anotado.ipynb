```python
def escuela_de_magia():     # Crea la función que incluye el juego 

    # INTRODUCCIÓN
    
    print ("Bienvenid@ pequeño ser de luz✨, ha llegado la convocatoria oficial de los Exámenes Mágicos Supremos📝, ¿deseas presentarte?")
    
    # Bucle que se repite hasta que el jugador responda "1" (continuar) o "2" (salir):
    
    while True:
        respuesta = input(" Escribe 1 para continuar✅, escribe 2 para salir del juego❌:")
        #respuesta = int(respuesta)  ----> Pensamos en tratar el 1 y 2 como caracteres numéricos pero decidimos mantenerlos como strings
        
        #Solo se aceptan "1" o "2", cualquier otra respuesta hace que vaya al else del final ("No entiendo tu respuesta, vuelve a intentarlo")
        if respuesta == "1" or respuesta == "2":
            if respuesta == "1":
                # El jugador ha elegido continuar por tanto salen la historia y las reglas del juego
                print("\nSabia decisión, el rector👴🏿 te leerá las reglas de los exámenes a continuación:\n")
                print("Eres un pequeño mago/a del reino de Alfea🪷, acabas de cumplir 18 años y te has matriculado en la Escuela Superior de Magia.\nDurante 3 años te has estado formando para convertirte en un/a gran mago/a de tu reino.\nEstás en tus exámenes finales, dependiendo de la nota que saques al final te graduarás como: Brujo/a, Hadito/a o Cambiaformas")

                
                # NIVEL 1: EXAMEN DE POCIONES
                # Hay que completar la receta con los ingredientes correctos
                
                print("\n\n\033[1m-----NIVEL 1: EXAMEN DE POCIONES-----\033[0m. \n\nEl primer examen es de pociones. Lo realizarás en el invernadero de la Bruja Suprema Asterin Blackbeak🧙🏼‍♀️. \nTendrás que elegir los ingredientes correctos para elaborar la poción para borrar recuerdos🔮.\n\n")
                print("Aquí tienes la receta incompleta, faltan 2 ingredientes: \n\n- Agua de luna🌙\n________________\n- Lágrimas de Fénix🐦‍🔥\n__________________\n- Escamas de serpiente dorada🐍\n\n")
                print("Estas son las opciones que tienes para completarla:\n1) Raíz de la amnesia🪾\n2) Polvo de Cronos🕐\n3) Plumas de grifo🪶\n4) Seda del Reino de las Arañas🕷️\n\nIndica el número del ingrediente correcto sin repetir.\n")
                
                
                while True:
                    lista_intentos_escobas = []             # Aquí es donde se guardan las 2 respuestas
                    puntuacion_e = 0                        # Contador para guardar los puntos (escobas) obtenidos en el nivel 1
                    import re                               # Para usar expresiones regulares (función re.search())
                    
                    intento_escoba1 = input("Indica el número correspondiente al primer ingrediente1️⃣:")     # Elige el primer ingrediente
                    if intento_escoba1 in "1,2,3,4":                                                         # Obliga a que sea 1,2,3 o 4
                        
                        # Si el primer ingrediente es 1,2,3 o 4, elige segundo ingrediente:
    
                        while True:
                            intento_escoba2 = input("Indica el número correspondiente al segundo ingrediente2️⃣:")
                            if intento_escoba2 in "1,2,3,4":
                                
                                # Si el segundo ingrediente también es 1,2,3 o 4, se guardan los dos intentos en la lista y se evalúan:
                                lista_intentos_escobas += [intento_escoba1,intento_escoba2]
                                
                                # Condiciones para la evaluación. Los break salen del bucle una vez evaluado:
                                if re.search(r"1",lista_intentos_escobas[0]) and re.search(r"2",lista_intentos_escobas[1]):             # Primero 1, segundo 2 (ambos correctos)
                                    puntuacion_e=puntuacion_e+3                                                                         # Suma 3 al contador de escobas
                                    print("\n\033[1mBIEN, tienes 3/3 escobas🧹🧹🧹\033[0m\n")                                          # Informa resultado
                                    break
                                elif re.search(r"2",lista_intentos_escobas[0]) and  re.search(r"1",lista_intentos_escobas[1]):          # Primero 2, segundo 1 (ambos correctos)
                                    puntuacion_e=puntuacion_e+3                                                                         # Suma 3 al contador de escobas
                                    print("\n\033[1mBIEN, tienes 3/3 escobas🧹🧹\033[0m\n")                                            # Informa resultado
                                    break
                                elif re.search(r"1|2",lista_intentos_escobas[0]) and  re.search(r"[^12]",lista_intentos_escobas[1]):    # Primero 1 o 2 (bien), segundo otro (3 o 4, mal)
                                    puntuacion_e=puntuacion_e+2                                                                         # Suma 2 al contador de escobas
                                    print("\n\033[1mCASI, tienes 2/3 escobas🧹🧹\033[0m\n")                                            # Informa resultado
                                    break
                                elif re.search(r"1|2",lista_intentos_escobas[1]) and  re.search(r"[^12]",lista_intentos_escobas[0]):    # Primero otro (3 o 4, mal), segundo 1 o 2 (bien)
                                    puntuacion_e=puntuacion_e+2                                                                         # Suma 2 al contador de escobas
                                    print("\n\033[1mCASI, tienes 2/3 escobas🧹🧹\033[0m\n")                                            # Informa resultado
                                    break
                                elif re.search(r"1",lista_intentos_escobas[1]) and  re.search(r"1",lista_intentos_escobas[0]):          # 1 en ambos intentos ---> cuenta una vez bien solo
                                    puntuacion_e=puntuacion_e+2                                                                         # Suma 2 al contador de escobas
                                    print("\n\033[1mCASI, tienes 2/3 escobas🧹🧹\033[0m\n")                                            # Informa resultado
                                    break
                                elif re.search(r"2",lista_intentos_escobas[1]) and  re.search(r"2",lista_intentos_escobas[0]):          # 2 en ambos intentos ---> cuenta una vez bien solo
                                    puntuacion_e=puntuacion_e+2                                                                         # Suma 2 al contador de escobas
                                    print("\n\033[1mCASI, tienes 2/3 escobas🧹🧹\033[0m\n")                                            # Informa resultado
                                    break
                                    #Si los dos ingredientes son incorrectos (3 o 4) recibe 1 punto 
                                elif re.search(r"3|4",lista_intentos_escobas[1]) and  re.search(r"3|4",lista_intentos_escobas[0]):      # 3 o 4 en ambos intentos (ambos mal)
                                    puntuacion_e=puntuacion_e+1                                                                         # Suma 1 al contador de escobas
                                    print("\n\033[1mOh vaya, tienes solo 1/3 escobas🧹\033[0m\n")                                      # Informa resultado
                                    break
                                #else:                                                    # Si pones cualquier cosa que no sea 1,2,3 o 4 contemplados en las condiciones
                                    #print("\nNo te entiendo pequeño ser de luz.🤨\n")      Como ya obligamos al principio y si no no deja avanzar, este ya no es necesario
                                    #break
                            else:
                                print("\nNo te entiendo pequeño ser de luz.🤨\n")         # El segundo ingrediente no es válido se vuelve a pedir
                        break                                                             # Ha terminado el examen por tanto sale del bucle del nivel 1 
                    else: 
                        print("\nNo te entiendo pequeño ser de luz.🤨\n")                 #El primer ingrediente no es válido, se repite

                
               
                # NIVEL 2: EXAMEN DE HECHIZOS
                # Hay que ordenar 3 palabras de un hechizo sin repetir
                    
                print("\n\n\033[1m-----NIVEL 2: EXAMEN DE HECHIZOS-----\033[0m \n\nEl segundo examen es sobre hechizos, con tu profesor Paladium Everwood, el duende🧝🏼‍♂️ que imparte sus clases en el Jardín de los Secretos.\nTendrás que conjurar un hechizo 🪄para teletransportarte 🌀.\nTienes que rellenar las 3 palabras que faltan:")
                print("\n“Donde —----- No Quiero —---, Donde —---- No Puedo Llegar. Transitus Liminis”\n\n")
                print("Estas son las opciones que tienes para completarla:\n1) Estar \n2) Quiero \n3) Estoy\n\nIndica el número de la palabra correcta sin repetir.\n")
                while True:
                    lista_intentos_hechizo = []                               # Guarda respuestas del jugador en este nivel
                    puntuacion_v = 0                                          # Cuenta puntos del nivel 2 (varitas)
                    
                    intento_hechizo1 = input("Indica el número correspondiente a la primera palabra1️⃣:")   # Elige primera palabra
                    if intento_hechizo1 in "1,2,3":                                                        # Obliga a que sea 1,2 o 3

                        # Si la primera palabra es 1,2 o 3, elige segunda palabra:
                        
                        while True:
                            intento_hechizo2 = input("Indica el número correspondiente a la segunda palabra2️⃣:")
                            if intento_hechizo2 in "1,2,3":

                                # Si la segunda palabra es 1,2 o 3, elige tercera palabra. Los break salen del bucle del intento 3:
                                
                                while True: 
                                    intento_hechizo3 = input("Indica el número correspondiente a la tercera palabra3️⃣:")
                                    if intento_hechizo3 in "1,2,3":
                                        
                                        # Si las tercera palabra también es 1,2 o 3, añade los tres intentos a la lista y evalúa:                                        
                                        lista_intentos_hechizo += [intento_hechizo1,intento_hechizo2,intento_hechizo3]
                                        
                                        #Ha contestado poniendo las 3 en su sitio (3,1,2) entonces ecibe 3 puntos
                                        if re.search(r"3",lista_intentos_hechizo[0]) and re.search(r"1",lista_intentos_hechizo[1]) and re.search(r"2",lista_intentos_hechizo[2]):
                                            puntuacion_v=puntuacion_v+3
                                            print("\n\033[1mBIEN, tienes 3/3 varitas🪄🪄🪄\033[0m")
                                            break
                                            
                                        #Acierta la primera y la segunda pero falla la tercera --> recibe 2 puntos 
                                        elif re.search(r"3",lista_intentos_hechizo[0]) and  re.search(r"1",lista_intentos_hechizo[1]) and re.search(r"[^2]",lista_intentos_hechizo[2]):
                                            puntuacion_v=puntuacion_v+2
                                            print("\n\033[1mCASI, tienes 2/3 varitas🪄🪄\033[0m")
                                            break
                                        
                                        #Acierta la primera y la tercera pero falla la segunda --> recibe 2 puntos
                                        elif re.search(r"3",lista_intentos_hechizo[0]) and  re.search(r"[^1]",lista_intentos_hechizo[1]) and re.search(r"2",lista_intentos_hechizo[2]):
                                            puntuacion_v=puntuacion_v+2
                                            print("\n\033[1mCASI, tienes 2/3 varitas🪄🪄\033[0m")
                                            break
                                        
                                        #Acierta la segunda y la tercera pero falla la primera --> recibe 2 puntos
                                        elif re.search(r"[^3]",lista_intentos_hechizo[0]) and  re.search(r"1",lista_intentos_hechizo[1]) and re.search(r"2",lista_intentos_hechizo[2]):
                                            puntuacion_v=puntuacion_v+2
                                            print("\n\033[1mCASI, tienes 2/3 varitas🪄🪄\033[0m")
                                            break
                                            
                                        # Por si pone algo que no sea 1,2 o 3, pero ya obligamos al principio, así que no hace falta
                                        #elif re.search(r"[^123]",lista_intentos_hechizo[0]) and  re.search(r"[^123]",lista_intentos_hechizo[1]) and re.search(r"[^123]",lista_intentos_hechizo[2]):
                                            #print("\n\033[1mNo entiendo.\033[0m")
                                            #break
                                            
                                        #Si pone cualquier otra combinación recibe 1 punto  
                                        else:
                                            puntuacion_v=puntuacion_v+1
                                            print("\n\033[1mOh vaya, tienes solo 1/3 varitas🪄\033[0m")
                                            break   
                                    else: 
                                        print("\nNo te entiendo pequeño ser de luz🤨.\n")    # Pone algo distinto de 1,2 o 3 en la tercera palabra, sigue pidiéndolo
                                break                                                        # Sale del bucle del intento 2
                            else:
                                print("\nNo te entiendo pequeño ser de luz🤨.\n")            # Pone algo distinto de 1,2 o 3 en la segunda palabra, sigue pidiéndolo
                        break                                                                # Sale del bucle del nivel 2 
                    else:
                        print("\nNo te entiendo pequeño ser de luz🤨.\n")                    # Pone algo distinto de 1,2 o 3 en la primera palabra, sigue pidiéndolo


                        
                # NIVEL 3: EXAMEN DE TRANSFORMACIONES
                # Consta de 3 misiones (A,B,C) para las que hay que transformarse en la criatura correcta 
                    
                print("\n\n\033[1m-----NIVEL 3: EXAMEN DE TRANSFORMACIONES-----\033[0m. \n\nEn último examen tienes que decidir en qué te transformarás para llevar a cabo distintas misiones. \nEste examen lo harás con tu profesora cambiaformas Violet Ennar🐆.")
                print("\nTendrás que transformarte en la criatura más adecuada para llevar a cabo cada misión. \nLas misiones son las siguientes:\n A) Llevar un mensaje cifrado al reino vecino📜. \n B) Encontrar las perlas escondidas en el fondo del Lago de las Sirenas🫧. \n C) Espiar al Capitán de la Guardia real💂🏻‍♀️.")
                print("\nA continuación aparecen tus opciones:\n 1) Cuervo🦅\n 2) Ratón🐀\n 3) Dragón marino🐲\n\nIndica el número de la criatura correcta sin repetir.\n")
                while True:
                    lista_intentos_camaleon = []                          # Respuestas del jugador 
                    puntuacion_c = 0                                      # Cuenta puntos del nivel 3 (camaleones)
                    
                    intento_camaleon1 = input("Indica el número correspondiente a la criatura para la misión A:")   # Elige criatura para misión A
                    if intento_camaleon1 in "1,2,3":                                                                # Obliga a que sea 1,2 o 3

                        # Si primer intento es 1,2 o 3, elige criatura para la misión B:
                        
                        while True:
                            intento_camaleon2 = input("Indica el número correspondiente a la criatura para la misión B:")
                            if intento_camaleon2 in "1,2,3":

                                # Si segundo intento es 1,2 o 3, elige criatura para la misión C. Los break salen del bucle del intento 3:
                                
                                while True:
                                    intento_camaleon3 = input("Indica el número correspondiente a la criatura para la misión C:")
                                    if intento_camaleon3 in "1,2,3":

                                        # Si tercer intento también es 1,2 o 3, añade intentos a la lista y evalúa:
                                        lista_intentos_camaleon += [intento_camaleon1,intento_camaleon2,intento_camaleon3]
                                        
                                        #Si responde a las 3 misiones de forma correcta (1,3,2) obtiene 3 puntos 
                                        if re.search(r"1",lista_intentos_camaleon[0]) and re.search(r"3",lista_intentos_camaleon[1]) and re.search(r"2",lista_intentos_camaleon[2]):
                                            puntuacion_c=puntuacion_c+3
                                            print("\n\033[1mBIEN, tienes 3/3 camaleones🦎🦎🦎\033[0m")
                                            break
                                            
                                        #Si acierta A y B pero falla C, obtiene 2 puntos 
                                        elif re.search(r"1",lista_intentos_camaleon[0]) and  re.search(r"3",lista_intentos_camaleon[1]) and re.search(r"[^2]",lista_intentos_camaleon[2]):
                                            puntuacion_c=puntuacion_c+2
                                            print("\n\033[1mCASI, tienes 2/3 camaleones🦎🦎\033[0m")
                                            break
                                            
                                        #Si acierta A y C pero falla B, obtiene 2 puntos
                                        elif re.search(r"1",lista_intentos_camaleon[0]) and  re.search(r"[^3]",lista_intentos_camaleon[1]) and re.search(r"2",lista_intentos_camaleon[2]):
                                            puntuacion_c=puntuacion_c+2
                                            print("\n\033[1mCASI, tienes 2/3 camaleones🦎🦎\033[0m")
                                            break
                                            
                                        #Si acierta B y C pero falla A, obtiene 2 puntos 
                                        elif re.search(r"[^1]",lista_intentos_camaleon[0]) and  re.search(r"3",lista_intentos_camaleon[1]) and re.search(r"2",lista_intentos_camaleon[2]):
                                            puntuacion_c=puntuacion_c+2
                                            print("\n\033[1mCASI, tienes 2/3 camaleones🦎🦎\033[0m")
                                            break
                                            
                                        # Si pone algo distinto de 1,2 o 3, pero ya se obliga al principio, no hace falta aquí
                                        #elif re.search(r"[^123]",lista_intentos_hechizo[0]) and  re.search(r"[^123]",lista_intentos_hechizo[1]) and re.search(r"[^123]",lista_intentos_hechizo[2]):
                                            #print("\nNo entiendo.🤨")
                                            #break

                                        # Si pone cualquier otra combinación obtiene 1 punto 
                                        else: 
                                            puntuacion_c=puntuacion_c+1
                                            print("\n\033[1mOh vaya, tienes solo 1/3 camaleones🦎")
                                            break
                                     
                                    else:
                                        print("\nNo te entiendo pequeño ser de luz.🤨\n")    # La criatura de la misión C no es 1,2 o 3; sigue pidiéndola
                                break                                                        # Sale del bucle de la misión B
                            else:
                                print("\nNo te entiendo pequeño ser de luz.🤨\n")            # La criatura de la misión B no es 1,2 o 3; sigue pidiéndola
                        break                                                                # Sale del bucle del nivel 3
                    else:
                        print("\nNo te entiendo pequeño ser de luz.🤨\n")                    # La criatura de la misión A no es 1,2 o 3; sigue pidiéndola
                        

                        
                #FIN DE CURSO: RESULTADO FINAL       
 
                print("\n\n\033[1m-----FIN DE CURSO-----\033[1m\n\nFelicidades, has terminado tus exámenes🎉🎊. Los maestros se reunirán y decidirán tu destino.\nSi has obtenido más escobas🧹 te graduarás como bruj@🧙🏻‍♀️.\nSi has obtenido más varitas🪄 te graduarás como hadit@🧚🏼‍♀️.\nSi has obtenido más camaleones🦎te graduarás como cambiaformas🪅.\nSi no alcanzas una nota igual o superior a 5 entre todos, debes repetir el curso🔙.\n")
                from IPython.display import Image, display               # Para mostar las imágenes en Jupyter 
                while True:
                    nota = puntuacion_e + puntuacion_v + puntuacion_c    # La nota final es igual a la suma de las tres puntuaciones
                    
                    if nota >= 5:                                        # Está aprobado si tiene 5 puntos o más
                        
                        if puntuacion_e > puntuacion_v and puntuacion_e > puntuacion_c:              # Si mayoría escobas
                            print("\n\033[1m¡ENHORABUENA! Te has convertido en bruj@\033[0m🧙🏻‍♀️.")    # Eres bruj@
                            display (Image("HOCUS-POCUS.webp", width=250))                          # Muestra imagen de brujas
                            break                                                                   # Sale del bucle de fin de curso
                            
                        elif puntuacion_v > puntuacion_e and puntuacion_v > puntuacion_c:           #Si mayoría varitas
                            print("\n\033[1m¡ENHORABUENA! Te has convertido en hadit@\033[0m🧚🏼‍♀️.")   # Eres hadit@
                            display (Image("Clarion.jpg", width=125))                               # Muestra imagen de hada
                            break                                                                   # Sale del bucle de fin de curso
                            
                        elif puntuacion_c > puntuacion_e and puntuacion_c > puntuacion_v:                  # Si mayoría camaleones
                            print("\n\033[1m¡ENHORABUENA! Te has convertido en cambiaformas\033[0m🪅.")   # Eres cambiaformas
                            display (Image("cambiaformas2.jpg", width=125))                                # Muestra imagen de cambiaformas
                            break                                                                          # Sale del bucle de fin de curso
                            
                        #Si hay un empate entre 2 puntuaciones o más, el jugador debe elegir un familiar:    
                        elif puntuacion_v == puntuacion_e or puntuacion_v == puntuacion_c or puntuacion_e == puntuacion_c:
                            print("\nVaya, parece que tienes dotes magníficas de varias disciplinas🤭 y los maestros no lo tienen claro.\nPara salir de dudas, elige un familiar🫶🏻 que te acompañará el resto de tu vida profesional mágica.\nTus poderes se mezclarán con los de tu familiar ¡Elige con cabeza! :\n\n1) Gato negro🐈‍⬛\n2) Búho🦉\n3) Serpiente alada🐍.\n")
                            while True:
                                familiar = input("Indica el número del familiar elegido: ")                        # Elige familiar
                                
                                if familiar == "1":                                                               # Si eliges 1 (gato)
                                    print("\n\033[1m¡ENHORABUENA! Te has convertido en bruj@🧙🏻‍♀️\033[0m.")          # Eres bruj@
                                    display (Image("HOCUS-POCUS.webp", width=250))                                # Muestra imagen de brujas
                                    break                                                                         # Sale del bucle de familiar
                                
                                elif familiar == "2":                                                             # Si eliges 2 (búho)
                                    print("\n\033[1m¡ENHORABUENA! Te has convertido en hadit@🧚🏼‍♀️\033[0m.")        # Eres hadit@
                                    display (Image("Clarion.jpg", width=125))                                     # Muestra imagen de hada
                                    break                                                                         # Sale del bucle de familiar
                                
                                elif familiar == "3":                                                             # Si eliges 3 (serpiente)
                                    print("\n\033[1m¡ENHORABUENA! Te has convertido en cambiaformas🪅\033[0m.")   # Eres cambiaformas
                                    display (Image("cambiaformas2.jpg", width=125))                               # Muestra imagen de cambiaformas
                                    break                                                                         # Sale del bucle de familiar
                                
                                else:                            # Si pone una opción no válida (distinto de 1,2 o 3), se vuelve a preguntar
                                    print("\nNo te entiendo🤨")
                                    
                    else:                                        # Si suspende (obtiene menos de 5 puntos) --> no se gradúa
                        print("Vaya, parece que aun no estás preparado para graduarte. Vuelve a intentarlo el curso que viene🧐.")
                        break                                    # Sale del bucle de fin de curso
                    break                                        # Sale del bucle principal, fin del juego
                    
            # El jugador ha elegido salir (ha elegido como respuesta el 2)        
            else:
                print ("Vaya!, parece que aun no estás preparad@, nos veremos en la próxima convocatoria➰")
                break                                            # Sale del bucle de fin de curso
            break                                                # Sale del bucle pincipal, fin del juego
            
        else:                                                        # La respuesta a si deseas presentarte es distinto de 1 o 2 --> sigue pidiéndola
            print("\nNo entiendo tu respuesta, vuelve a intentarlo🤨\n")               
```
